```python
import numpy as np

# Hours studied
X = np.array([1, 2, 3, 4, 5])

# Exam scores
y = np.array([30, 40, 50, 60, 70])
```

```python
X_b = np.c_[np.ones(len(X)),X]
print(X_b)
```

```python
teta = np.linalg.inv(X_b.T @ X_b) @X_b.T @y
```

```python
teta[0]
```

```python
teta[1]
```

```python
#Make prediction 
m= 5
prediction =  teta[0] +teta[1]*m
print(prediction)
```

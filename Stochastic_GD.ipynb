```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Data
X = np.array([1, 2, 2, 3, 3, 4, 4, 5], dtype=float)

# Prices in units of €100,000
y = np.array([
    1.2,
    1.8,
    2.0,
    2.6,
    2.8,
    3.5,
    3.7,
    4.5
], dtype=float) 
```

```python
#Parameter
teta0=0.0
teta1=0.0
alpha =0.01
epochs =100
m=len(X)
```

```python
for epcoh in range(epochs):
    indices = np.random.permutation(m)
    # print(indices)
    X_shuffled = X[indices]
    Y_shuffled = y[indices]

    # Go through ONE example at a time
    for i in range(m):
        #One Traing Example 
        x_i = X_shuffled[i]
        y_i = Y_shuffled[i]


        #Prediction
        prediction = teta0+teta1 * x_i

        #Error 
        error = prediction- y_i

        #Gradient 
        gradient_teta0 = error

        gradient_teta1  = error * x_i

        #Update Teta
        teta0 = teta0 -alpha * gradient_teta0
        teta1 = teta1 - alpha * gradient_teta1

          

```

```python
# ==========================================
# 4. FINAL PARAMETERS
# ==========================================

print("\nFinal theta0:", teta0)
print("Final theta1:", teta1)
```

```python


bedrooms = 4

predicted_price = teta0 + teta1 * bedrooms

# Convert back to euros
predicted_price_euros = predicted_price * 100000

print(
    f"Predicted price for {bedrooms} bedrooms: "
    f"€{predicted_price_euros:,.0f}"
)

```

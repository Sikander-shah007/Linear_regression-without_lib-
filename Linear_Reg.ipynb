```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1. DATA

# Number of bedrooms
X = np.array([1, 2, 2, 3, 3, 4, 4, 5], dtype=float)

# House prices in €100,000
# 1.2 means €120,000
# 4.5 means €450,000
y = np.array([1.2, 1.8, 2.0, 2.6, 2.8, 3.5, 3.7, 4.5], dtype=float)
```

```python
#intial values
#First we will give the values like tita0,tita1 and m = number of example 
teta0 =0.0
teta1 =0.0
m = len(X)
iteration = 100
alpha = 0.09
```

```python
#Now we will make the compute the cost how much error is their 
def compute_cost(X,Y,teta0,teta1):
    prediction = teta0+teta1*X
    # print(f'\n Predictoin : ${prediction} \n')
    error = prediction - Y
    # print(f'\n Error: ,${error}, \n')
    cost= (1/(2 * m))*np.sum(error**2)
    return cost

    
```

```python
#Now we will calculate the Teita values 
for i in range(iteration):
    prediction = teta0+teta1*X
    error = prediction-y

    gradient_teta0 = (1/m)*np.sum(error)
    gradient_teta1 = (1/m)*np.sum(error*X)

    #Update Tetas 
    teta0 = teta0 - alpha*gradient_teta0
    teta1 = teta1- alpha*gradient_teta1

    #Calculate Cost
    costE = compute_cost(X,y,teta0,teta1)






    
```

```python
print(f'Final Teta0 {teta0}')
```

```python
print(f'Final Teta1 {teta1}')
```

```python
# ==========================================
# 6. PREDICT HOUSE PRICE
# ==========================================

bedrooms = 4

predicted_price = teta0 + teta1 * bedrooms

print(
    f"Predicted price for {bedrooms} bedrooms: "
    f"€{predicted_price*100000 :,.0f}"
)
```

```python
plt.scatter(X, y, label="Actual houses")

x_line = np.linspace(1, 5,100)


y_line = teta0 + teta1 * x_line

plt.plot(x_line, y_line, label="Regression line")

plt.xlabel("Number of Bedrooms")
plt.ylabel("House Price (€100,000)")
plt.title("House Price Prediction")

plt.legend()
plt.show()
```

```python
import math
```

```python
# Training data
X = [1, 2, 2, 3, 3, 4, 4, 5]          # bedrooms
Y = [120, 180, 200, 260, 280, 350, 370, 450]  # price in €1000
```

```python
# House we want to predict
x = 2.5
```

```python
# Controls how much nearby points matter
tau = 1.0
```

```python
weights =[]
for xi in X:
    dis = (xi-x)
    w =math.exp(
        -(dis**2)/(2*tau**2))
    weights.append(w)
```

```python
weights
```

```python
numerator = 0
denominator = 0
for i in X:
    numerator =     numerator+ weights[i]*Y[i]
    denominator =  denominator + weights[i]

prediction = numerator/denominator



```
